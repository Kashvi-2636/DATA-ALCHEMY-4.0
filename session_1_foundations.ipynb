```python
!pip install -q groq
import time

from groq import Groq

API_KEY = None
try:
    from google.colab import userdata
    API_KEY = userdata.get("GROQ_API_KEY")
except Exception:
    API_KEY = None
if not API_KEY:
    import getpass
    API_KEY = getpass.getpass("Paste your Groq API key and press Enter: ")

client = Groq(api_key=API_KEY)

MODEL = "openai/gpt-oss-120b"

def call_llm(prompt, system=None, retries=3):

    messages = []

    if system:
        messages.append({"role": "system", "content": system})

    messages.append({"role": "user", "content": prompt})

    for attempt in range(retries):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                messages=messages
            )
            return resp.choices[0].message.content

        except Exception as e:
            if attempt == retries - 1:
                raise

            wait = 3 * (attempt + 1)
            print("Groq is busy, retrying in", wait, "seconds...")
            time.sleep(wait)
```

```python
print(call_llm("Does India has one of the wonders of the world and what is it?"))
```

```python
system = "You are Atlas, a friendly study buddy for a first-year student. Keep answers short and encouraging."
print(call_llm("I have 3 exams next week and I am panicking. What do I do?", system=system))
```

```python
system_a = "You are Atlas, a friendly study buddy for a first-year student. Keep answers short and encouraging."
system_b = "You are Atlas, give exam advice like a strict senior who has no patience."
question = "I have 3 exams next week and I am panicking. What do I do?"

print("--- Version A ---")
print(call_llm(question, system=system_a))
print("\n--- Version B ---")
print(call_llm(question, system=system_b))
```

```python
system = "You are Atlas. Before answering, think very carefully and step-by-step and list the plan in innovative way like experienced person."
print(call_llm("Help me plan a simple 3-day revision schedule for Maths, Physics and Chemistry.", system=system))
```

```python
question = "I have 3 exams next week and I am panicking. What do I do?"
personas = {
    "Strict coach": "You are Atlas, a strict coach. No excuses, give direct orders.",
    "Comedian": "You are Atlas, a comedian who cannot resist a joke, but still gives real advice.",
    "Pirate": "You are Atlas, who talks like a pirate, but still gives real advice.",
}

for name, system in personas.items():
    print(f"--- {name} ---")
    print(call_llm(question, system=system))
    print()
```
